```python
# Binary Activation Function
def step(x):
    return 1 if x >= 0 else 0

# Test
print(step(1))
print(step(-2))
```

```python
# Sigmoid Activation Function
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Test
print(sigmoid(0))
print(sigmoid(2))
```

```python
# Tanh Activation Function
import math

def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))

# Test
print(tanh(0))
print(tanh(1))
```

```python
# Relu AF
def relu(x):
    return max(0, x)

# Test
print(relu(5))
print(relu(-5))
```

```python
# Leaky Relu Activation Function
def leaky_relu(x, alpha=0.01):
    return x if x > 0 else alpha * x

# Test
print(leaky_relu(-5))
print(leaky_relu(3))
```

```python
# Softmax Activation Function
import math

def softmax(x):
    exp_values = [math.exp(i) for i in x]
    total = sum(exp_values)
    return [i / total for i in exp_values]

# Test
print(softmax([2.0, 1.0, 0.1]))
```

```python
# AF using Numpy
import numpy as np

def sigmoid_np(x):
    return 1 / (1 + np.exp(-x))

def relu_np(x):
    return np.maximum(0, x)

x = np.array([-2, -1, 0, 1, 2])
print(sigmoid_np(x))
print(relu_np(x))
```
